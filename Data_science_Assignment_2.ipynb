```python
# 1. Create a list of 5 integers. Perform and print the result of the following operations:
# append, extend, insert, remove, pop, clear, index, count, sort, and reverse.
numbers = [10, 25, 5, 40, 25]
print("Initial list:", numbers)

numbers.append(50)
print("After append(50):", numbers)

numbers.extend([60, 70])
print("After extend([60, 70]):", numbers)

numbers.insert(2, 15)
print("After insert(2, 15):", numbers)

numbers.remove(40)
print("After remove(40):", numbers)

popped_val = numbers.pop()
print(f"After pop() [removed {popped_val}]:", numbers)

idx = numbers.index(25)
print("Index of first 25:", idx)

cnt = numbers.count(25)
print("Count of 25:", cnt)

numbers.sort()
print("After sort():", numbers)

numbers.reverse()
print("After reverse():", numbers)

numbers.clear()
print("After clear():", numbers)
```

```python
# Section C: Exploring Pandas

# 1. Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
# - Use .info(), .describe()
# - Select 'Name' and 'City' columns
# - Drop 'City' column
# - Fill any missing values in 'Salary' column with the mean
# - Remove any duplicate rows

data = {
    "Name": [
        "Alice",
        "Bob",
        "Charlie",
        "David",
        "Eve",
        "Frank",
        "Grace",
        "Hank",
        "Alice",
        "Ivy",
    ],
    "Age": [25, 30, 35, 40, 22, 28, 32, 45, 25, 29],
    "City": [
        "NY",
        "LA",
        "Chicago",
        "Houston",
        "NY",
        "LA",
        "Chicago",
        "Houston",
        "NY",
        "LA",
    ],
    "Salary": [
        70000,
        80000,
        np.nan,
        90000,
        50000,
        60000,
        np.nan,
        100000,
        70000,
        65000,
    ],
}

df = pd.DataFrame(data)

print("--- Data Info ---")
df.info()

print("\n--- Data Summary Description ---")
print(df.describe())

print("\n--- Selecting Name and City ---")
print(df[["Name", "City"]])

df_dropped_city = df.drop(columns=["City"])
print("\n--- After Dropping City Column ---")
print(df_dropped_city)

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print("\n--- After Filling Missing Salary Values with Mean ---")
print(df)

df = df.drop_duplicates()
print("\n--- After Removing Duplicate Rows ---")
print(df)
```
