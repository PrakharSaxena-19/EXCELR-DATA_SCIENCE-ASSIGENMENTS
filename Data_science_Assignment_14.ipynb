```python
print(df.isnull().sum())
```

```python
X = df.drop("Type", axis=1)
y = df["Type"]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=X_pca[:,0],
    y=X_pca[:,1],
    hue=y,
    palette="Set1"
)

plt.title("PCA (2 Principal Components)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```

1. Missing values were checked before preprocessing.
2. Features were standardized using StandardScaler.
3. PCA reduced the dataset from multiple features to 2 principal components.
4. The scatter plot shows a clear separation among different wine classes.

### 1.
- PCA successfully reduced the dimensionality of the Wine dataset while preserving most of the important information.
- K-Means clustering was applied to both the original dataset and the PCA-transformed dataset.
- PCA improved data visualization by reducing the data to two principal components.
- The clustering results after PCA were similar to the original data, showing that important patterns were retained.

### 2. 
- PCA helps reduce the number of features, making data analysis faster and easier.
- It removes redundant information and reduces computational complexity.
- Clustering helps identify similar groups in unlabeled data and is useful for customer segmentation, image analysis, and pattern recognition.
- Combining PCA with clustering often improves visualization and reduces processing time.

### 3.
- Use PCA when the dataset has many correlated features or high dimensionality.
- Use clustering when the goal is to discover hidden groups or patterns in the data.
- Apply PCA before clustering when the dataset is large, as it reduces noise and computation while maintaining important information.
- For datasets with a small number of features, clustering can be applied directly without PCA.
