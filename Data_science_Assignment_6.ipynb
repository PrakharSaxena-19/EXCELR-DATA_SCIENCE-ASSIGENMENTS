***Assignment 6: Estimation, Confidence Intervals & Hypothesis Testing***

```python
import math
import numpy as np
import scipy.stats as stats
import pandas as pd
```

# **SECTION 1: ESTIMATION AND CONFIDENCE INTERVALS**

```python
# Data provided
durability_data = np.array(
    [
        1.13,
        1.55,
        1.43,
        0.92,
        1.25,
        1.36,
        1.32,
        0.85,
        1.07,
        1.48,
        1.20,
        1.33,
        1.18,
        1.22,
        1.29,
    ]
)

n = len(durability_data)
sample_mean = np.mean(durability_data)
sample_std = np.std(durability_data, ddof=1)  # Sample standard deviation (s)

print("--- SECTION 1: PRINT-HEAD DURABILITY ANALYSIS ---")
print(f"Sample Size (n): {n}")
print(f"Sample Mean (x̄): {sample_mean:.4f} million characters")
print(f"Sample Standard Deviation (s): {sample_std:.4f} million characters\n")
```

# **Confidence Interval Using Sample Standard Deviation**

```python
confidence_level_a = 0.99
alpha_a = 1 - confidence_level_a
df = n - 1

t_critical = stats.t.ppf(1 - alpha_a / 2, df)
margin_of_error_t = t_critical * (sample_std / math.sqrt(n))
ci_t = (sample_mean - margin_of_error_t, sample_mean + margin_of_error_t)

print("Task a: 99% Confidence Interval (Unknown σ, Sample Std Dev used)")
print(f"Degrees of Freedom: {df}")
print(f"t-Critical Value (t_0.005, 14): {t_critical:.4f}")
print(f"Margin of Error: ±{margin_of_error_t:.4f}")
print(
    f"99% Confidence Interval: ({ci_t[0]:.4f}, {ci_t[1]:.4f}) million characters\n"
)

pop_std = 0.2

z_critical = stats.norm.ppf(1 - alpha_a / 2)
margin_of_error_z = z_critical * (pop_std / math.sqrt(n))
ci_z = (sample_mean - margin_of_error_z, sample_mean + margin_of_error_z)

print("Task b: 99% Confidence Interval (Known σ = 0.2)")
print(f"z-Critical Value (z_0.005): {z_critical:.4f}")
print(f"Margin of Error: ±{margin_of_error_z:.4f}")
print(
    f"99% Confidence Interval: ({ci_z[0]:.4f}, {ci_z[1]:.4f}) million characters\n"
)
```

**SECTION 2: HYPOTHESIS TESTING (BOMBAY HOSPITALITY LTD.)**

```python
print("--- SECTION 2: HYPOTHESIS TESTING ---")
n_h = 25  # Sample size
sample_mean_cost = 3050  # x̄
mu_X = 600  # Mean units produced
sigma_X = 25  # Std dev of units produced

# Theoretical cost model: W = 1000 + 5*X
mu_theoretical = 1000 + 5 * mu_X  # μ = 1000 + 5*(600) = 4000
sigma_W = 5 * sigma_X  # σ_W = 5 * 25 = 125
alpha_h = 0.05  # Significance level

# 2. Calculate the Test Statistic (z):
z_stat = (sample_mean_cost - mu_theoretical) / (sigma_W / math.sqrt(n_h))

# 3. Determine Probability & Critical Value:
z_critical_h = stats.norm.ppf(1 - alpha_h)  # Right-tailed critical value
p_value = 1 - stats.norm.cdf(z_stat)

print(f"Theoretical Mean Operating Cost (μ): Rs. {mu_theoretical}")
print(f"Sample Mean Operating Cost (x̄): Rs. {sample_mean_cost}")
print(f"Standard Error (σ / √n): {sigma_W / math.sqrt(n_h)}")
print(f"Calculated Test Statistic (z): {z_stat:.4f}")
print(f"Critical Z-Value at α=0.05 (Right-tailed): {z_critical_h:.4f}")
print(f"P-value: {p_value:.4f}\n")
```
